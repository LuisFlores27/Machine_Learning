## Carga del DataSet

Posteriormente, se carga el archivo `titanic.csv`, el cual contiene la información del dataset que será utilizado para entrenar y probar el modelo de clasificación.

## Visualización de los datos

Finalmente, se muestran las primeras filas del dataset para verificar que los datos fueron cargados correctamente y observar la estructura de la información.

```python
# Importar la librería NumPy
import numpy as np

# Importar la librería Pandas
import pandas as pd

# Leer el archivo CSV y guardarlo en la variable df
df = pd.read_csv('titanic.csv')

# Mostrar las primeras 5 filas del dataset
df.head()
```

## Verificación de valores nulos

Antes de entrenar un modelo de Machine Learning como kNN, es necesario revisar si el DataSet contiene valores nulos o faltantes.

Los valores nulos representan información incompleta dentro del conjunto de datos y pueden provocar errores o afectar negativamente el rendimiento del modelo.

El algoritmo kNN calcula distancias entre registros para realizar clasificaciones, por lo que necesita trabajar con datos completos y correctamente estructurados.

En este paso se identificarán las columnas que contienen valores faltantes y la cantidad de registros nulos presentes en cada una.

```python
# Verificar la cantidad de valores nulos
# presentes en cada columna del DataSet
df.isnull().sum()
```

## Reemplazo de valores nulos en la columna Age

Después de identificar los valores faltantes dentro del DataSet, es necesario realizar un proceso de limpieza de datos.

En este caso, la columna `Age` contiene valores nulos. Para evitar problemas durante el entrenamiento del modelo kNN, dichos valores serán reemplazados utilizando el promedio de edades disponibles en el DataSet.

Este proceso se conoce como imputación de datos y permite conservar los registros sin necesidad de eliminarlos.

Posteriormente, se mostrarán los primeros 10 valores de la columna `Age` para verificar que el reemplazo se realizó correctamente.

```python
# Reemplazar los valores nulos (NaN) de la columna 'Age'
# utilizando el promedio de edades de la misma columna
df['Age'] = df['Age'].replace(
    np.nan,              # Valor que se desea reemplazar
    df['Age'].mean()     # Nuevo valor: promedio de la columna Age
)

# Mostrar los primeros 10 valores de la columna 'Age'
# para verificar que los valores nulos fueron reemplazados
df['Age'][:10]
```

## Conversión de variables categóricas a valores numéricos

Los algoritmos de Machine Learning, como kNN, trabajan únicamente con datos numéricos.

Sin embargo, el DataSet Titanic contiene columnas categóricas, como la columna `Sex`, cuyos valores son texto (`male` y `female`).

Por esta razón, es necesario transformar dichos valores categóricos en valores numéricos para que el modelo pueda procesarlos correctamente.

En este paso:

- `male` será reemplazado por `0`
- `female` será reemplazado por `1`

Finalmente, se mostrarán las primeras filas del DataSet para verificar que la conversión se realizó correctamente.

```python
# Crear un diccionario para convertir los valores
# de la columna 'Sex' a valores numéricos
Gender = {
    'Sex': {
        'male': 0,      # Reemplazar 'male' por 0
        'female': 1     # Reemplazar 'female' por 1
    }
}

# Reemplazar los valores categóricos utilizando
# el diccionario definido anteriormente
df = df.replace(Gender)
df.head()
```

## Eliminación de columnas innecesarias

Antes de entrenar el modelo kNN, es importante eliminar las columnas que no aportan información relevante para la clasificación.

En este caso, las columnas:

- `Name`
- `PassengerId`
- `Ticket`

no contribuyen directamente al análisis predictivo, ya que contienen identificadores o información textual que no ayuda al modelo a encontrar patrones útiles.

Además, mantener columnas irrelevantes puede afectar el rendimiento y la precisión del algoritmo.

Por esta razón, dichas columnas serán eliminadas del DataSet.

Finalmente, se mostrarán nuevamente las primeras filas para verificar que las columnas fueron eliminadas correctamente.

```python
# Eliminar las columnas 'Name', 'PassengerId' y 'Ticket'
# del DataSet
df.drop(
    ['Name', 'PassengerId', 'Ticket'],  # Columnas a eliminar
    axis=1,                             # axis=1 indica columnas
    inplace=True                        # Modificar el DataFrame original
)

df.head()
```

## Separación de variables independientes y variable objetivo

En Machine Learning, es necesario dividir el DataSet en:

- Variables independientes (`X`)
- Variable objetivo (`y`)

Las variables independientes contienen la información que el modelo utilizará para aprender patrones y realizar predicciones.

La variable objetivo representa el resultado que se desea predecir.

En este caso:

- `Survived` será la variable objetivo (`y`)
- El resto de columnas serán las variables independientes (`X`)

El modelo kNN utilizará las variables independientes para intentar predecir si un pasajero sobrevivió o no.

```python
# Crear la variable X eliminando la columna 'Survived'
# X contiene las variables independientes (features)
X = df.drop('Survived', axis=1)

# Crear la variable y utilizando únicamente
# la columna 'Survived'
# y representa la variable objetivo (target)
y = df.Survived
```

## División del DataSet en entrenamiento y prueba

Antes de entrenar el modelo kNN, es necesario dividir el DataSet en dos grupos de datos:

- Datos de entrenamiento (`train`)
- Datos de prueba (`test`)

Los datos de entrenamiento se utilizan para que el modelo aprenda patrones y relaciones dentro de la información.

Los datos de prueba se utilizan para evaluar el rendimiento del modelo con información que no ha visto anteriormente.

Esta separación permite verificar qué tan bien puede generalizar el modelo al trabajar con nuevos datos.

En este caso:

- El 70% de los datos será utilizado para entrenamiento
- El 30% restante será utilizado para pruebas

Además, se utilizará una semilla (`random_state`) para asegurar que la división de datos sea siempre la misma cada vez que se ejecute el notebook.

```python
# Importar la función train_test_split
# desde sklearn.model_selection
from sklearn.model_selection import train_test_split

# Dividir el DataSet en datos de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(

    X,                  # Variables independientes
    y,                  # Variable objetivo

    test_size=0.3,      # 30% de los datos para pruebas

    random_state=1      # Semilla para obtener resultados reproducibles
)
```

## Creación y entrenamiento del modelo kNN

Después de preparar y dividir el DataSet, el siguiente paso es crear el modelo de clasificación utilizando el algoritmo k-Nearest Neighbors (kNN).

El algoritmo kNN funciona comparando la distancia entre registros para identificar cuáles son los vecinos más cercanos a un nuevo dato.

Posteriormente, el modelo asigna la clase más común entre dichos vecinos.

En este caso:

- Se utilizarán 3 vecinos cercanos (`k = 3`)
- El modelo será entrenado utilizando los datos de entrenamiento previamente separados

El proceso de entrenamiento permite que el modelo aprenda los patrones presentes en el DataSet.

```python
# Importar el algoritmo KNeighborsClassifier
# desde sklearn.neighbors
from sklearn.neighbors import KNeighborsClassifier

# Crear el modelo kNN
# n_neighbors=3 indica que el modelo utilizará
# los 3 vecinos más cercanos para clasificar
knn = KNeighborsClassifier(n_neighbors=3)

# Entrenar el modelo utilizando:
# X_train -> variables independientes de entrenamiento
# y_train -> variable objetivo de entrenamiento
knn.fit(X_train, y_train)
```

## Realización de predicciones

Después de entrenar el modelo kNN, el siguiente paso consiste en utilizarlo para realizar predicciones sobre los datos de prueba.

El modelo analizará cada registro contenido en `X_test` y calculará cuáles son los vecinos más cercanos utilizando el valor de `k` definido anteriormente.

Posteriormente, el algoritmo asignará una clasificación basándose en la clase más frecuente entre dichos vecinos.

Las predicciones generadas serán almacenadas en una nueva variable llamada `y_pred`.

```python
# Utilizar el modelo kNN entrenado para realizar
# predicciones sobre los datos de prueba
y_pred = knn.predict(X_test)
```

## Evaluación del modelo

Después de realizar las predicciones, es necesario evaluar el rendimiento del modelo kNN para determinar qué tan precisas fueron sus clasificaciones.

En este paso se utilizará la métrica `accuracy`, la cual mide el porcentaje de predicciones correctas realizadas por el modelo.

La exactitud se calcula comparando:

- Los valores reales (`y_test`)
- Las predicciones realizadas por el modelo (`y_pred`)

Mientras más alto sea el valor de accuracy, mejor será el desempeño del modelo de clasificación.

```python
# Importar métricas de evaluación desde sklearn.metrics
from sklearn.metrics import accuracy_score, confusion_matrix

# Calcular la exactitud del modelo comparando:
# y_test -> valores reales
# y_pred -> predicciones realizadas por el modelo
knn_accuracy = round(
    accuracy_score(y_test, y_pred),
    2
)

# Mostrar el porcentaje de exactitud del modelo
print("Accuracy:", knn_accuracy * 100)
```

## Matriz de confusión

Además de la exactitud (`accuracy`), otra forma importante de evaluar un modelo de clasificación es mediante la matriz de confusión.

La matriz de confusión permite observar de manera detallada:

- Cuántas predicciones fueron correctas
- Cuántas predicciones fueron incorrectas
- Qué clases fueron confundidas por el modelo

En problemas de clasificación binaria, como el caso Titanic:

- `0` representa pasajeros que no sobrevivieron
- `1` representa pasajeros que sobrevivieron

La matriz ayuda a analizar el comportamiento real del modelo kNN y detectar posibles errores de clasificación.

```python
# Mostrar la matriz de confusión comparando:
# y_test -> valores reales
# y_pred -> predicciones generadas por el modelo
print(confusion_matrix(y_test, y_pred))
```
